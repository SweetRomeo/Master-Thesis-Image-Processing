```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def load_img():
    img = cv2.imread("./images/bricks.jpg").astype(np.float32) / 255
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img
```

```python
# load_img()
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111)
    ax.imshow(img)
```

```python
img = load_img()
display_img(img)
```

```python
# Gamma correctionda değer 1'den büyük olursa görüntü karanlıklaşırken, 1'den küçük olması durumunda görüntü aydınlanır.
gamma = 1/4
```

```python
result = np.power(img, gamma)
display_img(result)
```

```python
img = load_img()
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img, text = 'bricks', org=(10, 600), fontFace=font, fontScale=10, color=(255, 0, 0), thickness=4)
display_img(img)
```
