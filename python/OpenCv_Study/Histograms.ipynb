```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors
%matplotlib inline
```

```python
dark_horse = cv2.imread("./images/horse.jpg")
show_horse = cv2.cvtColor(dark_horse, cv2.COLOR_BGR2RGB) # Converted to RGB for show

rainbow = cv2.imread("./images/rainbow.jpg")
show_rainbow = cv2.cvtColor(rainbow, cv2.COLOR_BGR2RGB)

blue_bricks = cv2.imread("./images/bricks.jpg")
show_bricks = cv2.cvtColor(blue_bricks, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(show_horse)
```

```python
plt.imshow(show_rainbow)
```

```python
plt.imshow(show_bricks)
```

```python
hist_values = cv2.calcHist([blue_bricks], channels=[0], mask = None, histSize=[256], ranges=[0, 256])
# OpenCV BGR
```

```python
hist_values.shape
```

```python
plt.plot(hist_values)
```

```python
hist_values = cv2.calcHist([dark_horse], channels=[0], mask = None, histSize=[256], ranges=[0, 256])
```

```python
plt.plot(hist_values)
```

```python
img = dark_horse
```

```python
colors = ('b', 'g', 'r')

for i, color in enumerate(colors):
    histr = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(histr, color= color)
    plt.xlim([0, 50])
    plt.ylim([0, 500000])

plt.title("Histogram For Horse")
```
